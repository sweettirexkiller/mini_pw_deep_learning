```python
# create your own dataset class
class myDataset(Dataset):
    def __init__(self, X, y):
        self.X = X
        self.y = y
        self.n = X.shape[0]
    def __len__(self):
        return self.n
    def __getitem__(self, i):
        return (i, {"X":self.X[i], "y":self.y[i]})
```

```python
ds2 = myDataset(X, y)
ds2[4]
```

## DataLoader

```python
from torch.utils.data import DataLoader
```

```python
dl = DataLoader(ds, batch_size=10)
it = iter(dl)
print(next(it))
print(next(it))
```
