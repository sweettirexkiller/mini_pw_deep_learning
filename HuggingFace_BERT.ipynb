### BERT (PyTorch)

```python
from transformers import AutoTokenizer, AutoModel
from pprint import pprint

checkpoint = 'bert-base-uncased'
text = "When did the decline of coal start? It started when natural gas took off that started to begin in (President George W.) Bushs administration."
```

```python
tokenizer = AutoTokenizer.from_pretrained(checkpoint)

encoded_input = tokenizer(text)
print(text)
pprint(encoded_input)
```

```python
encoded_input = tokenizer(text, return_tensors='pt')  # return torch tensors
print(text)
pprint(encoded_input)
print(encoded_input['input_ids'].device)
```

```python
# encoded tensors are on cpu by default, need to move to GPU
```

```python
# function to move dicts to current device
def dict_2_dev(d):
    for k in d.keys():
        if hasattr(d[k], "to"):
            d[k] = d[k].to(device)
    return d
```

```python
dict_2_dev(encoded_input)
print(encoded_input['input_ids'].device)
```

```python
model = AutoModel.from_pretrained(checkpoint, device_map=device)
output = model(**encoded_input)
```

```python
%timeit  model(**encoded_input)
```

```python
from torch import nn
```

```python
m_imdb = nn.Sequential(nn.Linear(768, 1))
```

```python
dl_train=DataLoader(TensorDataset(Features_train, torch.tensor(y_train)), batch_size=128)
dl_test=DataLoader(TensorDataset(Features_test, torch.tensor(y_test)), batch_size=128)
```

```python
# training loop
from tqdm.notebook import tqdm
import torch.optim as optim

def fit(net, train=dl_train, valid=dl_test, epochs=10,
        learning_rate=None):
    net = net.to(device)

    if learning_rate is None:
        optimizer = optim.Adam(net.parameters())
    else:
        optimizer = optim.Adam(net.parameters(), lr=learning_rate)
    loss = torch.nn.BCEWithLogitsLoss()
    for e in range(epochs):
        net.train()
        epoch_loss = 0
        for X, y in tqdm(train):
            optimizer.zero_grad()
            X = X.to(device)
            y = y.to(device)
            pred = net(X)
            l = loss(pred, y.unsqueeze(1).float())
            epoch_loss += l.item()
            l.backward()
            optimizer.step()
        #if valid is not None:
        #    print("Accuracy at epoch", e, ":", eval_net(net, valid))
```

```python
fit(m_imdb)
```

```python
# function to evaluate accuracy
def eval_net(net, d):
    net = net.to(device)
    net.eval()
    with torch.no_grad():
        acc_sum = 0
        n = 0
        for x, y in tqdm(d):
            pred = net(x.to(device))
            predicted = (1*(pred > 0)).cpu()
            acc_sum += (predicted==y.unsqueeze(1)).sum().item()
            n += x.shape[0]
    return acc_sum/n
```

```python
eval_net(m_imdb, dl_test)
```
