```python
import torch
from diffusers import FluxPipeline

# Check if Metal (mps) is available
device = "mps" if torch.backends.mps.is_available() else "cpu"

pipe = FluxPipeline.from_pretrained("black-forest-labs/FLUX.1-dev", torch_dtype=torch.bfloat16)
pipe.enable_model_cpu_offload()  # Save some VRAM by offloading the model to CPU. Remove this if you have enough GPU power

prompt = "Create a minimalist black-and-white vector illustration of a cheerful programmer sitting in front of a laptop displaying code (</>). The character has spiky hair, round glasses, and a wide smile. One hand holds a glass of a dark beverage, and the other holds a steaming e-cigarette. Use a clean, bold line-art style with an off-white background to convey a relaxed, quirky, and somewhat rebellious vibe typical of a 'pathological' yet passionate coder"
image = pipe(
    prompt,
    height=1024,
    width=1024,
    guidance_scale=3.5,
    num_inference_steps=50,
    max_sequence_length=512,
    use_auth_token="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",
    generator=torch.Generator(device).manual_seed(0)
).images[0]
image.save("flux-dev.png")
```
